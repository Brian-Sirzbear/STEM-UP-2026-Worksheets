```python
import json
import os
import time
import tempfile

# ---------------------------------------------
# CONFIG
# ---------------------------------------------

FILE = "student_library.json"

# ---------------------------------------------
# HELPERS
# ---------------------------------------------

def load_library():

    try:
        with open(FILE, "r") as f:
            data = json.load(f)

            # safety check
            if isinstance(data, list):
                return data

            return []

    except (FileNotFoundError, json.JSONDecodeError):
        return []


def save_library(data):

    os.makedirs(os.path.dirname(FILE) or ".", exist_ok=True)

    with tempfile.NamedTemporaryFile(
        mode="w",
        delete=False,
        dir=os.path.dirname(FILE) or ".",
        suffix=".tmp"
    ) as tmp:

        json.dump(data, tmp, indent=2)

        temp_name = tmp.name

    os.replace(temp_name, FILE)


def collect_inputs():

    collected = []

    for (
        identifier_w,
        role_w,
        type_w,
        status_w,
        assume_w
    ) in inputs_widgets:

        identifier = identifier_w.value.strip()

        # skip blank rows
        if not identifier:
            continue

        collected.append({
            "input": identifier,
            "represents": role_w.value.strip(),
            "type": type_w.value,
            "status": status_w.value,
            "assumed": assume_w.value.strip()
        })

    return collected


def validate_form():

    errors = []

    if w_struc.value == CODE_STRUCTURES[0]:
        errors.append("Please select a Structure Type.")

    if not w_name.value.strip():
        errors.append("General Structure is required.")

    if not w_description.value.strip():
        errors.append("Brief Description is required.")

    if not collect_inputs():
        errors.append("At least one input is required.")

    return errors


def build_entry():

    return {

        "Structure Type":
            w_struc.value,

        "General Structure":
            w_name.value.strip(),

        "Library":
            w_library.value.strip(),

        "Number of Arguments":
            w_num_inputs.value.strip(),

        "Brief Description":
            w_description.value.strip(),

        "Additional Details":
            w_additional.value.strip(),

        "Source":
            w_source.value.strip(),

        "Inputs":
            collect_inputs(),

        "Timestamp":
            time.strftime("%Y-%m-%d %H:%M:%S")
    }


def replace_existing(entries, new_entry):

    name = new_entry["General Structure"].lower()

    filtered = [
        e for e in entries
        if e.get("General Structure", "").lower() != name
    ]

    filtered.append(new_entry)

    return filtered


# ---------------------------------------------
# SUBMIT HANDLER
# ---------------------------------------------

def on_submit(_):

    with out_submit:

        clear_output()

        # VALIDATION
        errors = validate_form()

        if errors:

            display(HTML(
                "<br>".join(
                    f'<div style="color:#9E1B34;font-weight:bold;">⚠️ {e}</div>'
                    for e in errors
                )
            ))

            return

        # BUILD ENTRY
        entry = build_entry()

        # LOAD OLD DATA
        library = load_library()

        # UPDATE
        library = replace_existing(library, entry)

        # SAVE
        save_library(library)

        # SUCCESS
        display(HTML(f"""
        <div style="
            background:#F5F2EC;
            border:2px solid #9E1B34;
            border-radius:12px;
            padding:20px;
            max-width:650px;
            font-family:system-ui;
        ">

            <div style="
                font-size:22px;
                font-weight:700;
                color:#9E1B34;
                margin-bottom:10px;
            ">
                ✅ Entry Added
            </div>

            <div style="margin-bottom:8px;">
                <b>General Structure:</b>
                {entry["General Structure"]}
            </div>

            <div style="margin-bottom:8px;">
                <b>Structure Type:</b>
                {entry["Structure Type"]}
            </div>

            <div style="margin-bottom:8px;">
                <b>Library:</b>
                {entry["Library"]}
            </div>

            <div style="margin-bottom:8px;">
                <b>Description:</b>
                {entry["Brief Description"]}
            </div>

            <div style="
                margin-top:14px;
                font-size:13px;
                color:#777;
            ">
                Saved at {entry["Timestamp"]}
            </div>

        </div>
        """))


# ---------------------------------------------
# BUTTON
# ---------------------------------------------

btn_submit._click_handlers.callbacks.clear()

btn_submit.on_click(on_submit)

display(btn_submit, out_submit)
```
